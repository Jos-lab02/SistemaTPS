# KPI 1: Tasa de Rotación de Personal (Attrition Rate)

Este indicador mide el porcentaje de empleados que abandonaron la organización respecto al total de empleados registrados en la base de datos .
Se uso la siguinete formula para obtener el porcentaje de empleados que abandonaron la empresa
### Fórmula

$$
Attrition\ Rate=
\frac{\text{Número de empleados que abandonaron la empresa}}
{\text{Número total de empleados}}
\times 100
$$

### Justificación

Se utiliza esta fórmula porque el atributo **Attrition** identifica si un empleado dejó la empresa ("Yes") o continúa trabajando en ella ("No"). Al dividir el número de empleados que abandonaron la organización entre el total de empleados, se obtiene la proporción de rotación del personal expresada en porcentaje.

---

# KPI 2: Ingreso Promedio Mensual por Departamento

Este indicador permite conocer el salario promedio de los empleados dentro de cada departamento de la organización.

### Fórmula
$$
Ingreso\ Promedio=
\frac{\text{Suma de los ingresos mensuales del departamento}}
{\text{Número de empleados del departamento}}
$$

### Justificación

Se utiliza la media aritmética de la variable **MonthlyIncome** agrupada por **Department**, ya que el objetivo es determinar el nivel promedio de compensación económica en cada área funcional de la empresa y detectar posibles diferencias salariales entre departamentos.

---

# KPI 3: Índice de Balance Vida-Trabajo Promedio por Cargo

Este indicador evalúa el nivel promedio de equilibrio entre la vida laboral y personal para cada cargo dentro de la organización.

### Fórmula

$$
WorkLifeBalance\ Promedio=
\frac{\text{Suma de las calificaciones de WorkLifeBalance}}
{\text{Número de empleados del cargo}}
$$

### Justificación

Se emplea la media de la variable **WorkLifeBalance** agrupada por **JobRole** porque permite medir la percepción promedio del equilibrio vida-trabajo en cada puesto laboral. Valores más bajos pueden indicar sobrecarga laboral y posibles riesgos de deserción futura.

---

# KPI 4: Brecha de Estancamiento Laboral

Este indicador analiza la diferencia entre el tiempo que un empleado ha permanecido en su cargo actual y el tiempo transcurrido desde su última promoción.

$$
Brecha\ Promedio_{Departamento}=
\frac{\sum(YearsInCurrentRole-YearsSinceLastPromotion)}
{\text{Número de empleados del departamento}}
$$

### Justificación

Se utiliza esta diferencia porque el enunciado establece que existe estancamiento laboral cuando los años en el puesto actual superan el tiempo transcurrido desde la última promoción. Una brecha positiva puede indicar falta de crecimiento profesional, disminución de motivación y posibles problemas de retención de talento dentro de la organización.



```python
import pandas as pd
ruta = "/content/WA_Fn-UseC_-HR-Employee-Attrition.csv"
df = pd.read_csv(ruta)
print("Número de registros:", len(df))
print()


#  KPI 1 Tasa de Rotación de Personal (Attrition Rate)

total_empleados = len(df)

empleados_salieron = df[df["Attrition"] == "Yes"].shape[0]

attrition_rate = (empleados_salieron / total_empleados) * 100

print("KPI 1: Tasa de Rotación de Personal (Attrition Rate)")

print(f"Empleados que salieron: {empleados_salieron}")
print(f"Total empleados: {total_empleados}")
print(f"Tasa de rotación: {attrition_rate:.2f}%")
print()

# KPI 2: Ingreso Promedio Mensual por Departamento

ingreso_promedio_departamento = (
    df.groupby("Department")["MonthlyIncome"]
    .mean()
    .round(2)
)

print("KPI 2: Ingreso Promedio Mensual por Departamento")

for departamento, ingreso in ingreso_promedio_departamento.items():

    print(f"\nDepartamento: {departamento}")
    print(f"Ingreso promedio mensual: ${ingreso:,.2f}")

# KPI 3: Índice de Balance Vida-Trabajo (Work-Life Balance Promedio) por Cargo


wlb_promedio_cargo = (
    df.groupby("JobRole")["WorkLifeBalance"]
    .mean()
    .round(2)
)

print("\nKPI 3: Índice de Balance Vida-Trabajo (Work-Life Balance Promedio) por Cargo")

for cargo, promedio in wlb_promedio_cargo.items():

    print(f"\nCargo: {cargo}")
    print(f"Balance vida-trabajo promedio: {promedio}/4")
# KPI 4: Brecha de Estancamiento Laboral (Puesto Actual vs. Última Promoción)

df["BrechaEstancamiento"] = (
    df["YearsInCurrentRole"]
    - df["YearsSinceLastPromotion"]
)
brecha_promedio = (
    df["BrechaEstancamiento"]
    .mean()
)

print("\nKPI 4: Brecha de Estancamiento Laboral (Puesto Actual vs. Última Promoción)")

print(f"Brecha promedio: {brecha_promedio:.2f}")
print()

brecha_departamento = (
    df.groupby("Department")["BrechaEstancamiento"]
    .mean()
    .round(2)
)

print("Brecha promedio por departamento")

for departamento, brecha in brecha_departamento.items():
    print(f"{departamento}: {brecha} años")
```
